```python
import numpy as np
import pandas as pd
import scipy.constants as sc
import matplotlib.pyplot as plt
from pygmid import Lookup as lk
```

```python
gm = 20e-3
```

```python
l=0.70
```

```python
gm_id=np.array([4, 5, 6])
```

```python
id=gm/gm_id
```

```python
choice = 1  #start from 0
devices = ['nfet_03v3', 'pfet_03v3', 'nfet_05v0', 'pfet_05v0', 'nfet_06v', 'pfet_06v0', 'nfet_10v0', 'pfet_10v0']
fet = lk('./simulation/'+devices[choice]+'.mat')
```

```python
jd = fet.lookup('ID_W', GM_ID=gm_id, L=1)
```

```python
w=id/jd
```

```python
cgg_w = fet.lookup('CGG_W', GM_ID=gm_id, L=1)
```

```python
cgg=w*cgg_w
```

```python
ft=gm/cgg/2/np.pi
```

```python
df = pd.DataFrame([gm_id, id, jd, w, cgg, ft], ['gm_id', 'id', 'jd', 'w', 'cgg', 'ft'], columns=['option1', 'option2', 'option3']); df
```
